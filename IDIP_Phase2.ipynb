```python
# ============================================================
# 2.1 STATE SELECTION ENGINE
# ============================================================

# Prepare available states
available_states = sorted(
    state_disaster["state_name"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)


def idip_select_state(state_name):
    """
    Validate and retrieve a selected state
    from the validated IDIP state summary.
    """

    if not isinstance(state_name, str):
        raise TypeError(
            "State name must be provided as text."
        )

    state_name = state_name.strip()

    # Case-insensitive state matching
    matches = [
        state
        for state in available_states
        if state.lower() == state_name.lower()
    ]

    # Handle invalid state
    if not matches:
        return None

    selected_state = matches[0]

    selected_state_record = state_disaster[
        state_disaster["state_name"]
        == selected_state
    ].copy()

    return selected_state, selected_state_record
```

```python
# 2.2 STATE OVERVIEW INTELLIGENCE

def generate_state_overview(state_name):
    result = idip_select_state(state_name)
    if result is None:
        return None

    selected_state, state_record = result
    row = state_record.iloc[0]
    earthquake_events = int(row["earthquake_events"])
    flood_events = int(row["flood_events"])
    cyclone_events = int(row["cyclone_events"])
    total_events = int(row["total_events"])
    disaster_counts = {
        "Earthquake": earthquake_events,
        "Flood": flood_events,
        "Cyclone": cyclone_events
    }

    if total_events > 0:
        dominant_disaster = max(
            disaster_counts,
            key=disaster_counts.get
        )
    else:
        dominant_disaster = "No recorded events"

    overview = {
        "state_name": selected_state,
        "total_events": total_events,
        "earthquake_events": earthquake_events,
        "flood_events": flood_events,
        "cyclone_events": cyclone_events,
        "dominant_disaster": dominant_disaster
    }

    return overview
```

```python
# 2.3.1 EARTHQUAKE INTELLIGENCE

def generate_earthquake_intelligence(state_name):

    # Select earthquake records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Handle no recorded earthquake events
    if state_eq.empty:

        return {
            "event_count": 0,
            "average_magnitude": None,
            "maximum_magnitude": None,
            "average_depth_km": None,
            "maximum_depth_km": None,
            "highest_magnitude_event_date": None,
            "coordinate_coverage": 0.0
        }

    # Highest magnitude event
    max_mag_idx = state_eq["magnitude"].idxmax()

    max_mag_date = state_eq.loc[
        max_mag_idx,
        "date"
    ]

    # Coordinate coverage
    valid_coordinates = state_eq[
        state_eq["latitude"].notna()
        & state_eq["longitude"].notna()
    ]

    coordinate_coverage = (
        len(valid_coordinates)
        / len(state_eq)
        * 100
    )

    # Build intelligence output
    intelligence = {

        "event_count": len(state_eq),

        "average_magnitude": round(
            state_eq["magnitude"].mean(),
            3
        ),

        "maximum_magnitude": round(
            state_eq["magnitude"].max(),
            3
        ),

        "average_depth_km": round(
            state_eq["depth_km"].mean(),
            3
        ),

        "maximum_depth_km": round(
            state_eq["depth_km"].max(),
            3
        ),

        "highest_magnitude_event_date": max_mag_date,

        "coordinate_coverage": round(
            coordinate_coverage,
            2
        )
    }

    return intelligence
```

```python
# 2.3.2 FLOOD INTELLIGENCE

def generate_flood_intelligence(state_name):

    # Select flood records for the state
    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Handle no recorded flood events
    if state_flood.empty:

        return {
            "event_count": 0,
            "average_duration_days": None,
            "maximum_duration_days": None,
            "longest_event_date": None,
            "coordinate_coverage": 0.0
        }

    # Longest-duration event
    max_duration_idx = state_flood["duration_days"].idxmax()

    longest_event_date = state_flood.loc[
        max_duration_idx,
        "date"
    ]

    # Coordinate coverage
    valid_coordinates = state_flood[
        state_flood["latitude"].notna()
        & state_flood["longitude"].notna()
    ]

    coordinate_coverage = (
        len(valid_coordinates)
        / len(state_flood)
        * 100
    )

    # Build intelligence output
    intelligence = {

        "event_count": len(state_flood),

        "average_duration_days": round(
            state_flood["duration_days"].mean(),
            3
        ),

        "maximum_duration_days": round(
            state_flood["duration_days"].max(),
            3
        ),

        "longest_event_date": longest_event_date,

        "coordinate_coverage": round(
            coordinate_coverage,
            2
        )
    }

    return intelligence
```

```python
# 2.3.3 CYCLONE INTELLIGENCE

def generate_cyclone_intelligence(state_name):

    # Select cyclone records for the state
    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Handle no recorded cyclone events
    if state_cyclone.empty:

        return {
            "event_count": 0,
            "average_wind_speed_knots": None,
            "maximum_wind_speed_knots": None,
            "average_pressure_mb": None,
            "minimum_pressure_mb": None,
            "strongest_wind_event_date": None,
            "coordinate_coverage": 0.0
        }

    # Strongest-wind event
    max_wind_idx = state_cyclone["wind_speed_knots"].idxmax()

    strongest_wind_event_date = state_cyclone.loc[
        max_wind_idx,
        "date"
    ]

    # Coordinate coverage
    valid_coordinates = state_cyclone[
        state_cyclone["latitude"].notna()
        & state_cyclone["longitude"].notna()
    ]

    coordinate_coverage = (
        len(valid_coordinates)
        / len(state_cyclone)
        * 100
    )

    # Build intelligence output
    intelligence = {

        "event_count": len(state_cyclone),

        "average_wind_speed_knots": round(
            state_cyclone["wind_speed_knots"].mean(),
            3
        ),

        "maximum_wind_speed_knots": round(
            state_cyclone["wind_speed_knots"].max(),
            3
        ),

        "average_pressure_mb": round(
            state_cyclone["pressure_mb"].mean(),
            3
        ),

        "minimum_pressure_mb": round(
            state_cyclone["pressure_mb"].min(),
            3
        ),

        "strongest_wind_event_date": strongest_wind_event_date,

        "coordinate_coverage": round(
            coordinate_coverage,
            2
        )
    }

    return intelligence
```

```python
# 2.4 COMBINED DISASTER INTELLIGENCE

def generate_disaster_intelligence(state_name):

    earthquake_intelligence = generate_earthquake_intelligence(
        state_name
    )

    flood_intelligence = generate_flood_intelligence(
        state_name
    )

    cyclone_intelligence = generate_cyclone_intelligence(
        state_name
    )

    intelligence = {
        "state_name": state_name,
        "earthquake": earthquake_intelligence,
        "flood": flood_intelligence,
        "cyclone": cyclone_intelligence
    }

    return intelligence
```

```python
# 2.5 TEMPORAL INTELLIGENCE

def generate_temporal_intelligence(state_name):

    # Select records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Convert dates
    state_eq["date"] = pd.to_datetime(
        state_eq["date"],
        errors="coerce"
    )

    state_flood["date"] = pd.to_datetime(
        state_flood["date"],
        errors="coerce"
    )

    state_cyclone["date"] = pd.to_datetime(
        state_cyclone["date"],
        errors="coerce"
    )

    # Create annual event counts
    annual_counts = {}

    for df in [
        state_eq,
        state_flood,
        state_cyclone
    ]:

        if not df.empty:

            yearly = (
                df["date"]
                .dropna()
                .dt.year
                .value_counts()
            )

            for year, count in yearly.items():

                annual_counts[year] = (
                    annual_counts.get(year, 0)
                    + int(count)
                )

    # Determine most active year
    if annual_counts:

        most_active_year = max(
            annual_counts,
            key=annual_counts.get
        )

        most_active_year_events = annual_counts[
            most_active_year
        ]

    else:

        most_active_year = None
        most_active_year_events = 0

    # Create annual pattern table
    annual_event_pattern = (
        pd.DataFrame(
            list(annual_counts.items()),
            columns=["year", "total_events"]
        )
        .sort_values("year")
        .reset_index(drop=True)
    )

    # Most active month
    all_dates = pd.concat(
        [
            state_eq["date"],
            state_flood["date"],
            state_cyclone["date"]
        ],
        ignore_index=True
    ).dropna()

    if not all_dates.empty:

        month_counts = (
            all_dates
            .dt.month
            .value_counts()
        )

        most_active_month = int(
            month_counts.idxmax()
        )

    else:

        most_active_month = None

    # Build intelligence output
    intelligence = {

        "most_active_year": most_active_year,

        "most_active_year_events":
            most_active_year_events,

        "most_active_month":
            most_active_month,

        "annual_event_pattern":
            annual_event_pattern
    }

    return intelligence
```

```python
# 2.5.2 RECENT ACTIVITY

def generate_recent_activity(state_name):

    # Select records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Convert dates
    state_eq["date"] = pd.to_datetime(
        state_eq["date"],
        errors="coerce"
    )

    state_flood["date"] = pd.to_datetime(
        state_flood["date"],
        errors="coerce"
    )

    state_cyclone["date"] = pd.to_datetime(
        state_cyclone["date"],
        errors="coerce"
    )

    # Combine all disaster dates
    all_events = pd.concat(
        [
            state_eq[["date"]].assign(
                disaster="Earthquake"
            ),
            state_flood[["date"]].assign(
                disaster="Flood"
            ),
            state_cyclone[["date"]].assign(
                disaster="Cyclone"
            )
        ],
        ignore_index=True
    )

    all_events = all_events.dropna(
        subset=["date"]
    )

    # Handle no available dates
    if all_events.empty:

        return {
            "latest_recorded_date": None,
            "latest_recorded_disaster": None,
            "recent_event_count": 0
        }

    # Latest recorded event
    latest_idx = all_events["date"].idxmax()

    latest_date = all_events.loc[
        latest_idx,
        "date"
    ]

    latest_disaster = all_events.loc[
        latest_idx,
        "disaster"
    ]

    # Define recent period from latest available record
    latest_year = latest_date.year

    recent_events = all_events[
        all_events["date"].dt.year >= latest_year - 4
    ]

    # Build intelligence output
    intelligence = {

        "latest_recorded_date":
            latest_date.strftime("%Y-%m-%d"),

        "latest_recorded_disaster":
            latest_disaster,

        "recent_event_count":
            len(recent_events)
    }

    return intelligence
```

```python
# 2.6 SPATIAL INTELLIGENCE

def generate_spatial_intelligence(state_name):

    # Select records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Calculate coordinate coverage
    coordinate_summary = {}

    for disaster_name, df in {
        "earthquake": state_eq,
        "flood": state_flood,
        "cyclone": state_cyclone
    }.items():

        if df.empty:

            coordinate_summary[disaster_name] = {
                "available_coordinates": 0,
                "total_records": 0,
                "coordinate_coverage": 0.0
            }

            continue

        valid_coordinates = df[
            df["latitude"].notna()
            & df["longitude"].notna()
        ]

        coordinate_count = len(valid_coordinates)

        coordinate_coverage = (
            coordinate_count
            / len(df)
            * 100
        )

        coordinate_summary[disaster_name] = {
            "available_coordinates": coordinate_count,
            "total_records": len(df),
            "coordinate_coverage": round(
                coordinate_coverage,
                2
            )
        }

    # Build intelligence output
    intelligence = {
        "state_name": state_name,
        "geographic_distribution": coordinate_summary
    }

    return intelligence
```

```python
# 2.7 SEVERITY INTELLIGENCE

def generate_severity_intelligence(state_name):

    # Select records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Earthquake severity
    if not state_eq.empty:

        earthquake_severity = {
            "average_magnitude": round(
                state_eq["magnitude"].mean(),
                3
            ),
            "maximum_magnitude": round(
                state_eq["magnitude"].max(),
                3
            ),
            "average_depth_km": round(
                state_eq["depth_km"].mean(),
                3
            ),
            "maximum_depth_km": round(
                state_eq["depth_km"].max(),
                3
            )
        }

    else:

        earthquake_severity = {
            "average_magnitude": None,
            "maximum_magnitude": None,
            "average_depth_km": None,
            "maximum_depth_km": None
        }

    # Flood severity
    if not state_flood.empty:

        flood_severity = {
            "average_duration_days": round(
                state_flood["duration_days"].mean(),
                3
            ),
            "maximum_duration_days": round(
                state_flood["duration_days"].max(),
                3
            )
        }

    else:

        flood_severity = {
            "average_duration_days": None,
            "maximum_duration_days": None
        }

    # Cyclone severity
    if not state_cyclone.empty:

        cyclone_severity = {
            "average_wind_speed_knots": round(
                state_cyclone["wind_speed_knots"].mean(),
                3
            ),
            "maximum_wind_speed_knots": round(
                state_cyclone["wind_speed_knots"].max(),
                3
            ),
            "average_pressure_mb": round(
                state_cyclone["pressure_mb"].mean(),
                3
            ),
            "minimum_pressure_mb": round(
                state_cyclone["pressure_mb"].min(),
                3
            )
        }

    else:

        cyclone_severity = {
            "average_wind_speed_knots": None,
            "maximum_wind_speed_knots": None,
            "average_pressure_mb": None,
            "minimum_pressure_mb": None
        }

    # Build intelligence output
    intelligence = {

        "state_name": state_name,

        "earthquake": earthquake_severity,

        "flood": flood_severity,

        "cyclone": cyclone_severity
    }

    return intelligence
```

```python
# 2.8 HISTORICAL EVENT TIMELINE

def generate_historical_timeline(state_name):

    # Select earthquake records
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Select flood records
    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Select cyclone records
    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Convert dates
    state_eq["date"] = pd.to_datetime(
        state_eq["date"],
        errors="coerce"
    )

    state_flood["date"] = pd.to_datetime(
        state_flood["date"],
        errors="coerce"
    )

    state_cyclone["date"] = pd.to_datetime(
        state_cyclone["date"],
        errors="coerce"
    )

    # Create earthquake timeline
    earthquake_timeline = pd.DataFrame({
        "date": state_eq["date"],
        "disaster": "Earthquake",
        "severity": state_eq["magnitude"],
        "severity_metric": "Magnitude"
    })

    # Create flood timeline
    flood_timeline = pd.DataFrame({
        "date": state_flood["date"],
        "disaster": "Flood",
        "severity": state_flood["duration_days"],
        "severity_metric": "Duration (days)"
    })

    # Create cyclone timeline
    cyclone_timeline = pd.DataFrame({
        "date": state_cyclone["date"],
        "disaster": "Cyclone",
        "severity": state_cyclone["wind_speed_knots"],
        "severity_metric": "Wind Speed (knots)"
    })

    # Combine timelines
    timeline = pd.concat(
        [
            earthquake_timeline,
            flood_timeline,
            cyclone_timeline
        ],
        ignore_index=True
    )

    # Remove records without valid dates
    timeline = timeline.dropna(
        subset=["date"]
    )

    # Sort chronologically
    timeline = timeline.sort_values(
        "date"
    ).reset_index(drop=True)

    # Build intelligence output
    intelligence = {
        "state_name": state_name,
        "event_count": len(timeline),
        "timeline": timeline
    }

    return intelligence
```

```python
# 2.9 DATA RELIABILITY INTELLIGENCE

def generate_reliability_intelligence(state_name):

    # Select records for the state
    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    # Calculate coordinate coverage
    coordinate_coverage = {}

    for disaster_name, df in {
        "earthquake": state_eq,
        "flood": state_flood,
        "cyclone": state_cyclone
    }.items():

        if df.empty:

            coordinate_coverage[disaster_name] = None

            continue

        valid_coordinates = df[
            df["latitude"].notna()
            & df["longitude"].notna()
        ]

        coordinate_coverage[disaster_name] = round(
            len(valid_coordinates)
            / len(df)
            * 100,
            2
        )

    # Calculate severity availability
    severity_availability = {

        "earthquake": (
            state_eq["magnitude"].notna().sum()
            if not state_eq.empty
            else 0
        ),

        "flood": (
            state_flood["duration_days"].notna().sum()
            if not state_flood.empty
            else 0
        ),

        "cyclone": (
            state_cyclone["wind_speed_knots"].notna().sum()
            if not state_cyclone.empty
            else 0
        )
    }

    # Build data limitations
    data_limitations = []

    if not state_eq.empty:

        if coordinate_coverage["earthquake"] < 100:
            data_limitations.append(
                "Not all earthquake records contain coordinates."
            )

        if severity_availability["earthquake"] < len(state_eq):
            data_limitations.append(
                "Not all earthquake records contain magnitude."
            )

    if not state_flood.empty:

        if coordinate_coverage["flood"] < 100:
            data_limitations.append(
                "Not all flood records contain coordinates."
            )

        if severity_availability["flood"] < len(state_flood):
            data_limitations.append(
                "Not all flood records contain duration."
            )

    if not state_cyclone.empty:

        if coordinate_coverage["cyclone"] < 100:
            data_limitations.append(
                "Not all cyclone records contain coordinates."
            )

        if severity_availability["cyclone"] < len(state_cyclone):
            data_limitations.append(
                "Not all cyclone records contain wind speed."
            )

    if not data_limitations:

        data_limitations.append(
            "No field-level coverage limitations detected."
        )

    # Build intelligence output
    intelligence = {

        "state_name": state_name,

        "state_data_coverage": {
            "earthquake_records": len(state_eq),
            "flood_records": len(state_flood),
            "cyclone_records": len(state_cyclone)
        },

        "coordinate_coverage":
            coordinate_coverage,

        "severity_field_availability":
            severity_availability,

        "data_limitations":
            data_limitations
    }

    return intelligence
```

```python
# 2.10 STATE INTELLIGENCE CONSOLIDATION

def generate_state_intelligence(state_name):

    # Generate state overview
    overview = generate_state_overview(
        state_name
    )

    # Generate disaster intelligence
    disaster_intelligence = generate_disaster_intelligence(
        state_name
    )

    # Generate temporal intelligence
    temporal_intelligence = generate_temporal_intelligence(
        state_name
    )

    # Generate recent activity
    recent_activity = generate_recent_activity(
        state_name
    )

    # Generate spatial intelligence
    spatial_intelligence = generate_spatial_intelligence(
        state_name
    )

    # Generate severity intelligence
    severity_intelligence = generate_severity_intelligence(
        state_name
    )

    # Generate historical timeline
    historical_timeline = generate_historical_timeline(
        state_name
    )

    # Generate reliability intelligence
    reliability = generate_reliability_intelligence(
        state_name
    )

    # Build complete state intelligence
    state_intelligence = {

        "overview": overview,

        "disaster_intelligence":
            disaster_intelligence,

        "temporal_intelligence":
            temporal_intelligence,

        "recent_activity":
            recent_activity,

        "spatial_intelligence":
            spatial_intelligence,

        "severity_intelligence":
            severity_intelligence,

        "historical_timeline":
            historical_timeline,

        "reliability":
            reliability
    }

    return state_intelligence
```

# STATE INTELLIGENCE ENGINE

```python
# ============================================================
# 12.1 STATE DISASTER PROFILE
# ============================================================

def generate_state_disaster_profile(state_name):

    overview = generate_state_overview(
        state_name
    )

    if overview is None:
        return None

    total_events = overview["total_events"]
    earthquake_events = overview["earthquake_events"]
    flood_events = overview["flood_events"]
    cyclone_events = overview["cyclone_events"]
    dominant_disaster = overview["dominant_disaster"]

    if total_events == 0:

        profile = (
            f"{state_name} has no recorded disaster events "
            "in the current IDIP dataset."
        )

        return profile

    # Calculate event shares
    earthquake_share = (
        earthquake_events / total_events * 100
    )

    flood_share = (
        flood_events / total_events * 100
    )

    cyclone_share = (
        cyclone_events / total_events * 100
    )

    # Build narrative
    profile = (
        f"{state_name} has {total_events:,} recorded disaster "
        f"events in the current IDIP dataset. "
        f"Floods account for {flood_events:,} events "
        f"({flood_share:.1f}%), earthquakes account for "
        f"{earthquake_events:,} events "
        f"({earthquake_share:.1f}%), and cyclones account "
        f"for {cyclone_events:,} events "
        f"({cyclone_share:.1f}%). "
        f"The highest recorded event count is associated "
        f"with {dominant_disaster.lower()}."
    )

    return profile
```

```python
# ============================================================
# 12.2 DOMINANT DISASTER SUMMARY
# ============================================================

def generate_dominant_disaster_summary(state_name):

    overview = generate_state_overview(
        state_name
    )

    if overview is None:
        return None

    dominant_disaster = overview["dominant_disaster"]
    total_events = overview["total_events"]

    if total_events == 0:

        return (
            f"No dominant recorded disaster can be identified "
            f"for {state_name} because no events are recorded "
            "in the current IDIP dataset."
        )

    disaster_counts = {
        "Earthquake": overview["earthquake_events"],
        "Flood": overview["flood_events"],
        "Cyclone": overview["cyclone_events"]
    }

    dominant_count = disaster_counts[
        dominant_disaster
    ]

    dominant_share = (
        dominant_count
        / total_events
        * 100
    )

    summary = (
        f"{dominant_disaster} has the highest recorded event "
        f"count for {state_name}, with {dominant_count:,} "
        f"events representing {dominant_share:.1f}% of all "
        f"recorded disaster events in the current dataset."
    )

    return summary
```

```python
# ============================================================
# 12.3 HISTORICAL PATTERN SUMMARY
# ============================================================

def generate_historical_pattern_summary(state_name):

    temporal = generate_temporal_intelligence(
        state_name
    )

    if temporal is None:
        return None

    most_active_year = temporal["most_active_year"]
    most_active_year_events = temporal["most_active_year_events"]
    most_active_month = temporal["most_active_month"]

    if most_active_year is None:

        return (
            f"No historical temporal pattern could be identified "
            f"for {state_name} from the available dated records."
        )

    month_names = {
        1: "January",
        2: "February",
        3: "March",
        4: "April",
        5: "May",
        6: "June",
        7: "July",
        8: "August",
        9: "September",
        10: "October",
        11: "November",
        12: "December"
    }

    month_name = month_names.get(
        most_active_month,
        "an unknown month"
    )

    summary = (
        f"The highest recorded annual disaster activity "
        f"for {state_name} occurred in {most_active_year}, "
        f"with {most_active_year_events} recorded events. "
        f"{month_name} had the highest recorded event frequency "
        f"when all available disaster records are considered "
        f"together."
    )

    return summary
```

```python
# ============================================================
# 12.4 SEVERITY SUMMARY
# ============================================================

def generate_severity_summary(state_name):

    severity = generate_severity_intelligence(
        state_name
    )

    if severity is None:
        return None

    summary_parts = []

    # --------------------------------------------------------
    # Earthquake severity
    # --------------------------------------------------------

    earthquake = severity["earthquake"]

    if earthquake["maximum_magnitude"] is not None:

        summary_parts.append(
            f"Recorded earthquakes reached a maximum "
            f"magnitude of {earthquake['maximum_magnitude']:.1f}, "
            f"with an average magnitude of "
            f"{earthquake['average_magnitude']:.1f}."
        )

    # --------------------------------------------------------
    # Flood severity
    # --------------------------------------------------------

    flood_data = severity["flood"]

    if flood_data["maximum_duration_days"] is not None:

        summary_parts.append(
            f"Recorded floods had an average duration of "
            f"{flood_data['average_duration_days']:.1f} days "
            f"and a maximum recorded duration of "
            f"{flood_data['maximum_duration_days']:.1f} days."
        )

    # --------------------------------------------------------
    # Cyclone severity
    # --------------------------------------------------------

    cyclone_data = severity["cyclone"]

    if cyclone_data["maximum_wind_speed_knots"] is not None:

        summary_parts.append(
            f"Recorded cyclones reached a maximum wind speed "
            f"of {cyclone_data['maximum_wind_speed_knots']:.1f} "
            f"knots, with an average of "
            f"{cyclone_data['average_wind_speed_knots']:.1f} "
            f"knots."
        )

    # --------------------------------------------------------
    # No severity information
    # --------------------------------------------------------

    if not summary_parts:

        return (
            f"No usable severity information is available "
            f"for the recorded disaster events in {state_name}."
        )

    # --------------------------------------------------------
    # Combine severity statements
    # --------------------------------------------------------

    return " ".join(summary_parts)
```

```python
# ============================================================
# 12.5 DATA RELIABILITY STATEMENT
# ============================================================

def generate_data_reliability_statement(state_name):

    reliability = generate_reliability_intelligence(
        state_name
    )

    if reliability is None:
        return None

    statements = []

    # --------------------------------------------------------
    # Record coverage
    # --------------------------------------------------------

    coverage = reliability[
        "state_data_coverage"
    ]

    total_records = (
        coverage["earthquake_records"]
        + coverage["flood_records"]
        + coverage["cyclone_records"]
    )

    if total_records > 0:

        statements.append(
            f"IDIP contains {total_records:,} recorded "
            f"disaster events associated with {state_name} "
            f"across the available earthquake, flood and "
            f"cyclone datasets."
        )

    else:

        statements.append(
            f"No disaster records are currently available "
            f"for {state_name} in the IDIP datasets."
        )

    # --------------------------------------------------------
    # Coordinate coverage
    # --------------------------------------------------------

    coordinate_coverage = reliability[
        "coordinate_coverage"
    ]

    coordinate_statements = []

    for disaster, coverage_value in coordinate_coverage.items():

        if coverage_value is None:
            continue

        disaster_name = disaster.capitalize()

        if coverage_value == 100:

            coordinate_statements.append(
                f"{disaster_name} records have complete "
                f"coordinate coverage."
            )

        elif coverage_value == 0:

            coordinate_statements.append(
                f"{disaster_name} records have no available "
                f"coordinates."
            )

        else:

            coordinate_statements.append(
                f"{disaster_name} records have "
                f"{coverage_value:.1f}% coordinate coverage."
            )

    statements.extend(coordinate_statements)

    # --------------------------------------------------------
    # Data limitations
    # --------------------------------------------------------

    limitations = reliability[
        "data_limitations"
    ]

    if limitations:

        statements.append(
            "The following data limitations should be "
            "considered when interpreting the results: "
            + " ".join(limitations)
        )

    # --------------------------------------------------------
    # Final reliability statement
    # --------------------------------------------------------

    return " ".join(statements)
```

```python
# ============================================================
# 12.6 COMPLETE NARRATIVE INTELLIGENCE
# ============================================================

def generate_narrative_intelligence(state_name):

    # Generate narrative components
    state_profile = generate_state_disaster_profile(
        state_name
    )

    dominant_summary = generate_dominant_disaster_summary(
        state_name
    )

    historical_summary = generate_historical_pattern_summary(
        state_name
    )

    severity_summary = generate_severity_summary(
        state_name
    )

    reliability_statement = generate_data_reliability_statement(
        state_name
    )

    # Build narrative intelligence
    narrative = {

        "state_disaster_profile":
            state_profile,

        "dominant_disaster_summary":
            dominant_summary,

        "historical_pattern_summary":
            historical_summary,

        "severity_summary":
            severity_summary,

        "data_reliability_statement":
            reliability_statement
    }

    return narrative
```

```python
# ============================================================
# 13.1 FINAL STATE INTELLIGENCE REPORT
# ============================================================

def generate_final_state_report(state_name):

    # Generate complete state intelligence
    intelligence = generate_state_intelligence(
        state_name
    )

    # Generate narrative intelligence
    narrative = generate_narrative_intelligence(
        state_name
    )

    # Build final report
    final_report = {

        "state_name": state_name,

        "state_overview":
            intelligence["overview"],

        "disaster_breakdown":
            intelligence["disaster_intelligence"],

        "temporal_intelligence":
            {
                "pattern":
                    intelligence["temporal_intelligence"],
                "recent_activity":
                    intelligence["recent_activity"]
            },

        "spatial_intelligence":
            intelligence["spatial_intelligence"],

        "severity_intelligence":
            intelligence["severity_intelligence"],

        "historical_event_timeline":
            intelligence["historical_timeline"],

        "evidence_and_reliability":
            intelligence["reliability"],

        "narrative_intelligence":
            narrative
    }

    return final_report
```

```python
# ============================================================
# 16.1 TEMPORAL ACTIVITY VISUALIZATION
# ============================================================

def prepare_temporal_activity_data(state_name):

    temporal = generate_temporal_intelligence(
        state_name
    )

    if temporal is None:
        return None

    annual_pattern = temporal["annual_event_pattern"]

    if annual_pattern is None or annual_pattern.empty:
        return None

    temporal_data = annual_pattern.copy()

    temporal_data = temporal_data.sort_values(
        "year"
    ).reset_index(drop=True)

    return temporal_data
```

```python
# ============================================================
# 16.2 TEMPORAL ACTIVITY VISUALIZATION
# ============================================================

def plot_temporal_activity(state_name):

    temporal_data = prepare_temporal_activity_data(
        state_name
    )

    if temporal_data is None:
        return

    fig, ax = plt.subplots(figsize=(11, 5))

    ax.plot(
        temporal_data["year"],
        temporal_data["total_events"],
        marker="o",
        linewidth=2
    )

    ax.set_title(
        f"{state_name} — Recorded Disaster Activity Over Time",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Year")
    ax.set_ylabel("Recorded Events")

    ax.grid(
        True,
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 17.3 CYCLONE WIND SPEED VISUALIZATION
# ============================================================

def plot_cyclone_severity(state_name):

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        ==
        state_name.strip().lower()
    ].copy()

    state_cyclone = state_cyclone[
        state_cyclone["wind_speed_knots"].notna()
    ]

    if state_cyclone.empty:
        print(
            f"No usable cyclone wind-speed data available "
            f"for {state_name}."
        )
        return

    fig, ax = plt.subplots(figsize=(9, 5))

    ax.hist(
        state_cyclone["wind_speed_knots"],
        bins=10,
        edgecolor="black"
    )

    ax.set_title(
        f"{state_name} — Cyclone Wind Speed Distribution",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Wind Speed (knots)")
    ax.set_ylabel("Recorded Events")

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 18.1 SPATIAL VISUALIZATION DATA
# ============================================================

def prepare_spatial_data(state_name):

    result = {}

    # --------------------------------------------------------
    # Earthquake coordinates
    # --------------------------------------------------------

    state_earthquake = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    earthquake_points = state_earthquake.dropna(
        subset=["latitude", "longitude"]
    )

    result["earthquake"] = earthquake_points[
        ["latitude", "longitude"]
    ].copy()

    # --------------------------------------------------------
    # Flood coordinates
    # --------------------------------------------------------

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_name.strip().lower()
    ].copy()

    flood_points = state_flood.dropna(
        subset=["latitude", "longitude"]
    )

    result["flood"] = flood_points[
        ["latitude", "longitude"]
    ].copy()

    # --------------------------------------------------------
    # Cyclone coordinates
    # --------------------------------------------------------

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        ==
        state_name.strip().lower()
    ].copy()

    cyclone_points = state_cyclone.dropna(
        subset=["latitude", "longitude"]
    )

    result["cyclone"] = cyclone_points[
        ["latitude", "longitude"]
    ].copy()

    return result
```

```python
# ============================================================
# 18.2 STATE SPATIAL VISUALIZATION
# ============================================================

def plot_state_spatial_distribution(state_name):

    spatial_data = prepare_spatial_data(
        state_name
    )

    fig, ax = plt.subplots(
        figsize=(9, 9)
    )

    plotted = False

    # --------------------------------------------------------
    # Earthquake
    # --------------------------------------------------------

    earthquake_points = spatial_data["earthquake"]

    if not earthquake_points.empty:

        ax.scatter(
            earthquake_points["longitude"],
            earthquake_points["latitude"],
            s=35,
            alpha=0.7,
            label="Earthquake"
        )

        plotted = True

    # --------------------------------------------------------
    # Flood
    # --------------------------------------------------------

    flood_points = spatial_data["flood"]

    if not flood_points.empty:

        ax.scatter(
            flood_points["longitude"],
            flood_points["latitude"],
            s=35,
            alpha=0.7,
            label="Flood"
        )

        plotted = True

    # --------------------------------------------------------
    # Cyclone
    # --------------------------------------------------------

    cyclone_points = spatial_data["cyclone"]

    if not cyclone_points.empty:

        ax.scatter(
            cyclone_points["longitude"],
            cyclone_points["latitude"],
            s=35,
            alpha=0.7,
            label="Cyclone"
        )

        plotted = True

    # --------------------------------------------------------
    # No coordinates
    # --------------------------------------------------------

    if not plotted:

        ax.text(
            0.5,
            0.5,
            "No event coordinates available",
            ha="center",
            va="center",
            transform=ax.transAxes,
            fontsize=12
        )

    ax.set_title(
        f"{state_name} — Available Disaster Coordinates",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel(
        "Longitude (°E)"
    )

    ax.set_ylabel(
        "Latitude (°N)"
    )

    ax.grid(
        True,
        linestyle="--",
        alpha=0.4
    )

    if plotted:
        ax.legend()

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 21.1 DISASTER TREND COMPARISON
# ============================================================

def prepare_disaster_trend_data(state_name):

    state_key = state_name.strip().lower()

    # --------------------------------------------------------
    # Earthquake
    # --------------------------------------------------------

    eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    eq_trend = (
        eq.groupby("year")
        .size()
        .rename("Earthquake")
    )

    # --------------------------------------------------------
    # Flood
    # --------------------------------------------------------

    fl = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    fl_trend = (
        fl.groupby("year")
        .size()
        .rename("Flood")
    )

    # --------------------------------------------------------
    # Cyclone
    # --------------------------------------------------------

    cy = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    cy_trend = (
        cy.groupby("year")
        .size()
        .rename("Cyclone")
    )

    # --------------------------------------------------------
    # Combine
    # --------------------------------------------------------

    trend = pd.concat(
        [
            eq_trend,
            fl_trend,
            cy_trend
        ],
        axis=1
    ).fillna(0)

    # Create complete yearly sequence
    min_year = trend.index.min()
    max_year = trend.index.max()

    trend = trend.reindex(
        range(min_year, max_year + 1),
        fill_value=0
    )

    trend = trend.astype(int)

    trend.index.name = "year"

    return trend.reset_index()
```

```python
# ============================================================
# 21.2 DISASTER TREND VISUALIZATION
# ============================================================

def plot_disaster_trend(state_name):

    trend = prepare_disaster_trend_data(
        state_name
    )

    if trend.empty:
        print(
            f"No disaster trend data available "
            f"for {state_name}."
        )
        return

    fig, ax = plt.subplots(
        figsize=(11, 5)
    )

    ax.plot(
        trend["year"],
        trend["Earthquake"],
        marker="o",
        linewidth=2,
        label="Earthquake"
    )

    ax.plot(
        trend["year"],
        trend["Flood"],
        marker="o",
        linewidth=2,
        label="Flood"
    )

    ax.plot(
        trend["year"],
        trend["Cyclone"],
        marker="o",
        linewidth=2,
        label="Cyclone"
    )

    ax.set_title(
        f"{state_name} — Disaster Activity Over Time",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Year")
    ax.set_ylabel("Recorded Events")

    ax.grid(
        True,
        linestyle="--",
        alpha=0.4
    )

    ax.legend()

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 22.1 EARTHQUAKE MAGNITUDE PROFILE
# ============================================================

def plot_earthquake_severity(state_name):

    state_key = state_name.strip().lower()

    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    if state_eq.empty:
        print(
            f"No recorded earthquake events for {state_name}."
        )
        return

    # Convert dates safely
    state_eq["date"] = pd.to_datetime(
        state_eq["date"],
        errors="coerce",
        format="mixed"
    )

    # Remove records without usable dates or magnitude
    state_eq = state_eq.dropna(
        subset=["date", "magnitude"]
    ).sort_values("date")

    if state_eq.empty:
        print(
            f"No usable earthquake severity data "
            f"available for {state_name}."
        )
        return

    # Event positions
    event_positions = range(
        1,
        len(state_eq) + 1
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    ax.bar(
        list(event_positions),
        state_eq["magnitude"].astype(float)
    )

    for position, magnitude in zip(
        event_positions,
        state_eq["magnitude"]
    ):
        ax.text(
            position,
            float(magnitude) + 0.05,
            f"{magnitude:.1f}",
            ha="center",
            fontweight="bold"
        )

    ax.set_title(
        f"{state_name} — Recorded Earthquake Magnitudes",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Earthquake Event")
    ax.set_ylabel("Magnitude")

    # Date labels
    ax.set_xticks(
        list(event_positions)
    )

    ax.set_xticklabels(
        state_eq["date"].dt.strftime("%Y-%m-%d"),
        rotation=45,
        ha="right"
    )

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 22.2 FLOOD DURATION PROFILE
# ============================================================

def plot_flood_severity(state_name):

    state_key = state_name.strip().lower()

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    if state_flood.empty:
        print(
            f"No recorded flood events for {state_name}."
        )
        return

    # Remove records without valid duration
    state_flood["duration_days"] = pd.to_numeric(
        state_flood["duration_days"],
        errors="coerce"
    )

    state_flood = state_flood.dropna(
        subset=["duration_days"]
    )

    if state_flood.empty:
        print(
            f"No usable flood duration data "
            f"available for {state_name}."
        )
        return

    # Count events by duration
    duration_counts = (
        state_flood["duration_days"]
        .value_counts()
        .sort_index()
    )

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    bars = ax.bar(
        duration_counts.index.astype(str),
        duration_counts.values
    )

    ax.set_title(
        f"{state_name} — Recorded Flood Duration",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Duration (days)")
    ax.set_ylabel("Recorded Events")

    # Display event counts
    for bar, count in zip(
        bars,
        duration_counts.values
    ):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            count + 0.3,
            str(int(count)),
            ha="center",
            fontweight="bold"
        )

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 22.3 CYCLONE SEVERITY STATUS
# ============================================================

def get_cyclone_severity_status(state_name):

    state_key = state_name.strip().lower()

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ].copy()

    if state_cyclone.empty:
        return {
            "status": "NO_RECORDED_EVENTS",
            "message": (
                f"No recorded cyclone events available "
                f"for {state_name}."
            )
        }

    valid_wind = state_cyclone[
        state_cyclone["wind_speed_knots"].notna()
    ]

    if valid_wind.empty:
        return {
            "status": "NO_SEVERITY_DATA",
            "message": (
                f"Cyclone events are recorded for "
                f"{state_name}, but usable wind-speed "
                f"severity data is unavailable."
            )
        }

    return {
        "status": "AVAILABLE",
        "message": (
            f"Cyclone severity data available for "
            f"{state_name}."
        )
    }
```

```python
# ============================================================
# 24.1 DISASTER SHARE / COMPOSITION VISUALIZATION
# ============================================================

def plot_disaster_share(state_name):

    state_key = state_name.strip().lower()

    state_record = state_disaster[
        state_disaster["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ]

    if state_record.empty:
        print(f"State not found: {state_name}")
        return

    row = state_record.iloc[0]

    breakdown = pd.Series(
        {
            "Earthquake": int(row["earthquake_events"]),
            "Flood": int(row["flood_events"]),
            "Cyclone": int(row["cyclone_events"])
        }
    )

    total_events = breakdown.sum()

    if total_events == 0:
        print(
            f"No recorded disaster events available "
            f"for {state_name}."
        )
        return

    # Remove zero-event categories from the pie
    # so the visualization does not create misleading slices.
    breakdown = breakdown[
        breakdown > 0
    ]

    fig, ax = plt.subplots(
        figsize=(8, 6)
    )

    ax.pie(
        breakdown.values,
        labels=breakdown.index,
        autopct="%1.1f%%",
        startangle=90
    )

    ax.set_title(
        f"{state_name} — Recorded Disaster Composition",
        fontsize=14,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# 25.1 FINAL VISUALIZATION VALIDATION
# ============================================================

def validate_visualization_layer(state_name):

    state_key = state_name.strip().lower()

    # --------------------------------------------------------
    # State record
    # --------------------------------------------------------

    state_record = state_disaster[
        state_disaster["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ]

    if state_record.empty:
        raise ValueError(
            f"State not found: {state_name}"
        )

    row = state_record.iloc[0]

    expected_eq = int(row["earthquake_events"])
    expected_flood = int(row["flood_events"])
    expected_cyclone = int(row["cyclone_events"])

    # --------------------------------------------------------
    # Earthquake
    # --------------------------------------------------------

    state_eq = earthquake[
        earthquake["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ]

    actual_eq = len(state_eq)

    # --------------------------------------------------------
    # Flood
    # --------------------------------------------------------

    state_flood = flood[
        flood["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ]

    actual_flood = len(state_flood)

    # --------------------------------------------------------
    # Cyclone
    # --------------------------------------------------------

    state_cyclone = cyclone[
        cyclone["state_name"]
        .astype(str)
        .str.strip()
        .str.lower()
        == state_key
    ]

    actual_cyclone = len(state_cyclone)

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    checks = {

        "earthquake_count": (
            expected_eq == actual_eq
        ),

        "flood_count": (
            expected_flood == actual_flood
        ),

        "cyclone_count": (
            expected_cyclone == actual_cyclone
        ),

        "total_event_count": (
            int(row["total_events"])
            ==
            actual_eq
            + actual_flood
            + actual_cyclone
        )
    }

    print("=" * 70)
    print("IDIP — FINAL VISUALIZATION VALIDATION")
    print("=" * 70)

    print(f"State: {state_name}")
    print()

    print(
        f"Earthquake : {actual_eq} / {expected_eq}",
        "✓" if checks["earthquake_count"] else "✗"
    )

    print(
        f"Flood      : {actual_flood} / {expected_flood}",
        "✓" if checks["flood_count"] else "✗"
    )

    print(
        f"Cyclone    : {actual_cyclone} / {expected_cyclone}",
        "✓" if checks["cyclone_count"] else "✗"
    )

    print()

    if all(checks.values()):

        print(
            "FINAL VALIDATION: PASSED ✓"
        )

    else:

        print(
            "FINAL VALIDATION: FAILED ✗"
        )

        for name, result in checks.items():

            if not result:
                print(
                    f"  Failed check: {name}"
                )

    return checks
```

```python
# ============================================================
# 27.4 ML EVENT-LEVEL DATASET
# ============================================================

features = [
    "latitude",
    "longitude",
    "depth_km",
    "magnitude"
]

dedup_columns = [
    "year",
    "month",
    "latitude",
    "longitude",
    "depth_km",
    "magnitude"
]

ml_columns = [
    "event_id",
    "source",
    "date",
    "year",
    "month",
    "latitude",
    "longitude",
    "depth_km",
    "magnitude",
    "state_name",
    "district_name"
]

earthquake_ml = earthquake[
    [
        column
        for column in ml_columns
        if column in earthquake.columns
    ]
].copy()

# ------------------------------------------------------------
# Ensure numeric values
# ------------------------------------------------------------

for column in features:
    earthquake_ml[column] = pd.to_numeric(
        earthquake_ml[column],
        errors="coerce"
    )

# ------------------------------------------------------------
# Ensure date format
# ------------------------------------------------------------

earthquake_ml["date"] = pd.to_datetime(
    earthquake_ml["date"],
    errors="coerce"
)

# ------------------------------------------------------------
# Remove records unusable for ML
# ------------------------------------------------------------

earthquake_ml = earthquake_ml.dropna(
    subset=features
).copy()

# ------------------------------------------------------------
# Sort chronologically
# ------------------------------------------------------------

earthquake_ml = earthquake_ml.sort_values(
    ["date", "event_id"]
).reset_index(drop=True)

# ------------------------------------------------------------
# Event-level deduplication
# ------------------------------------------------------------

before_dedup = len(earthquake_ml)

earthquake_ml = earthquake_ml.drop_duplicates(
    subset=dedup_columns,
    keep="first"
).reset_index(drop=True)

after_dedup = len(earthquake_ml)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

remaining_duplicates = earthquake_ml.duplicated(
    subset=dedup_columns
).sum()

print("=" * 70)
print("IDIP — ML EVENT-LEVEL DATASET")
print("=" * 70)

print(f"Records before deduplication : {before_dedup}")
print(f"Records after deduplication  : {after_dedup}")
print(f"Duplicates removed           : {before_dedup - after_dedup}")
print(f"Remaining duplicates         : {remaining_duplicates}")
print()

print(
    f"Date range: "
    f"{earthquake_ml['date'].min()} → "
    f"{earthquake_ml['date'].max()}"
)
```

```python
# ============================================================
# 27.17 INDIA-BOUNDARY DISTANCE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# Build India's unified boundary
# ------------------------------------------------------------

india_union = india_boundary.geometry.union_all()

# Boundary only
india_boundary_line = india_union.boundary

# ------------------------------------------------------------
# Function: geodesic distance to India's boundary
# ------------------------------------------------------------

def distance_to_india_boundary_km(latitude, longitude):
    
    # Use a local Azimuthal Equidistant projection
    # centered on the earthquake
    local_crs = (
        f"+proj=aeqd "
        f"+lat_0={latitude} "
        f"+lon_0={longitude} "
        f"+datum=WGS84 "
        f"+units=m"
    )

    transformer = Transformer.from_crs(
        "EPSG:4326",
        local_crs,
        always_xy=True
    )

    # Transform candidate point
    x, y = transformer.transform(
        longitude,
        latitude
    )

    point = Point(x, y)

    # Transform India boundary into same local projection
    boundary_gdf = gpd.GeoSeries(
        [india_boundary_line],
        crs="EPSG:4326"
    )

    boundary_projected = boundary_gdf.to_crs(
        local_crs
    ).iloc[0]

    distance_km = (
        point.distance(boundary_projected)
        / 1000
    )

    return distance_km


# ------------------------------------------------------------
# Calculate distance for all 22 candidates
# ------------------------------------------------------------

candidate_distance = candidate_validation.copy()

candidate_distance[
    "distance_to_india_km"
] = candidate_distance.apply(
    lambda row:
        distance_to_india_boundary_km(
            row["latitude"],
            row["longitude"]
        ),
    axis=1
)

# ------------------------------------------------------------
# Sort nearest → farthest
# ------------------------------------------------------------

candidate_distance = (
    candidate_distance
    .sort_values(
        "distance_to_india_km"
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Preliminary relevance classification
# ------------------------------------------------------------

def classify_india_relevance(distance):

    if distance == 0:
        return "INSIDE INDIA"

    elif distance <= 100:
        return "CROSS-BORDER — HIGH RELEVANCE"

    elif distance <= 250:
        return "REGIONAL — MODERATE RELEVANCE"

    else:
        return "OUTSIDE INDIA — LOW STATE RELEVANCE"


candidate_distance[
    "india_relevance"
] = candidate_distance[
    "distance_to_india_km"
].apply(
    classify_india_relevance
)

print("=" * 70)
print("IDIP — INDIA BOUNDARY DISTANCE ANALYSIS")
print("=" * 70)

print(
    "Total ML candidates:",
    len(candidate_distance)
)

print()

display(
    candidate_distance[
        [
            "event_id",
            "latitude",
            "longitude",
            "magnitude",
            "depth_km",
            "anomaly_score",
            "distance_to_india_km",
            "india_relevance"
        ]
    ]
)
```

```python
# ============================================================
# 27.19 NEAREST INDIAN RELEVANT STATE
# ============================================================

# ------------------------------------------------------------
# 1. Create state-level geometries
# ------------------------------------------------------------

india_states = (
    india_boundary[
        ["st_nm", "st_code", "geometry"]
    ]
    .dissolve(
        by=["st_nm", "st_code"],
        as_index=False
    )
)

print("=" * 70)
print("IDIP — INDIA STATE GEOMETRY PREPARATION")
print("=" * 70)

print(
    "Indian states / territories:",
    len(india_states)
)

print(
    "CRS:",
    india_states.crs
)

# ------------------------------------------------------------
# 2. Create candidate point GeoDataFrame
# ------------------------------------------------------------

candidate_points = gpd.GeoDataFrame(
    candidate_distance.copy(),
    geometry=gpd.points_from_xy(
        candidate_distance["longitude"],
        candidate_distance["latitude"]
    ),
    crs="EPSG:4326"
)

# ------------------------------------------------------------
# 3. Project both datasets to a metric CRS
# ------------------------------------------------------------
#
# EPSG:6933 = World Equidistant Cylindrical
# Used here for consistent metric distance calculation.
#
# The resulting distance is an analytical proximity measure,
# not a geological impact estimate.
# ------------------------------------------------------------

metric_crs = "EPSG:6933"

candidate_metric = candidate_points.to_crs(
    metric_crs
)

states_metric = india_states.to_crs(
    metric_crs
)

# ------------------------------------------------------------
# 4. Find nearest Indian state
# ------------------------------------------------------------

nearest_state = gpd.sjoin_nearest(
    candidate_metric,
    states_metric[
        [
            "st_nm",
            "st_code",
            "geometry"
        ]
    ],
    how="left",
    distance_col="nearest_state_distance_m"
)

# ------------------------------------------------------------
# 5. Convert distance to kilometres
# ------------------------------------------------------------

nearest_state[
    "nearest_state_distance_km"
] = (
    nearest_state[
        "nearest_state_distance_m"
    ] / 1000
)

# ------------------------------------------------------------
# 6. Rename derived state field
# ------------------------------------------------------------

nearest_state = nearest_state.rename(
    columns={
        "st_nm": "relevant_state",
        "st_code": "relevant_state_code"
    }
)

# ------------------------------------------------------------
# 7. Remove spatial index column if present
# ------------------------------------------------------------

if "index_right" in nearest_state.columns:
    nearest_state = nearest_state.drop(
        columns=["index_right"]
    )

# ------------------------------------------------------------
# 8. Final ordering
# ------------------------------------------------------------

nearest_state = (
    nearest_state
    .sort_values("anomaly_score")
    .reset_index(drop=True)
)

print("=" * 70)
print("IDIP — NEAREST INDIAN RELEVANT STATE")
print("=" * 70)

print(
    "Candidates processed:",
    len(nearest_state)
)

print()

display(
    nearest_state[
        [
            "event_id",
            "latitude",
            "longitude",
            "magnitude",
            "depth_km",
            "anomaly_score",
            "distance_to_india_km",
            "india_relevance",
            "relevant_state",
            "nearest_state_distance_km"
        ]
    ]
)
```

```python
# ============================================================
# 27.23 FINAL ML INTEGRITY VALIDATION
# ============================================================

print("=" * 70)
print("IDIP — FINAL EARTHQUAKE ML INTEGRITY VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Final ML event count
# ------------------------------------------------------------

print("Unique ML events:", len(earthquake_ml))

# ------------------------------------------------------------
# 2. Duplicate check
# ------------------------------------------------------------

duplicate_count = earthquake_ml.duplicated(
    subset=dedup_columns
).sum()

print(
    "Remaining duplicate events:",
    duplicate_count
)

# ------------------------------------------------------------
# 3. Training / testing period validation
# ------------------------------------------------------------

print(
    "Training period:",
    train_data["year"].min(),
    "-",
    train_data["year"].max()
)

print(
    "Testing period:",
    test_data["year"].min(),
    "-",
    test_data["year"].max()
)

# ------------------------------------------------------------
# 4. Temporal leakage check
# ------------------------------------------------------------

temporal_leakage = (
    train_data["year"].max()
    >=
    test_data["year"].min()
)

print(
    "Temporal leakage:",
    temporal_leakage
)

# ------------------------------------------------------------
# 5. ML candidate period
# ------------------------------------------------------------

candidate_dates = pd.to_datetime(
    final_ml_output["date"],
    errors="coerce"
)

candidate_years = candidate_dates.dt.year

print(
    "ML candidate period:",
    candidate_years.min(),
    "-",
    candidate_years.max()
)

# ------------------------------------------------------------
# 6. Candidate count
# ------------------------------------------------------------

print(
    "ML anomaly candidates:",
    len(final_ml_output)
)

# ------------------------------------------------------------
# 7. Regional candidates
# ------------------------------------------------------------

regional_count = (
    final_ml_output[
        "state_regional_intelligence"
    ].sum()
)

print(
    "India regional candidates:",
    int(regional_count)
)

print(
    "Low state-relevance candidates:",
    len(final_ml_output)
    - int(regional_count)
)

# ------------------------------------------------------------
# 8. Coordinate integrity
# ------------------------------------------------------------

missing_coordinates = (
    final_ml_output[
        ["latitude", "longitude"]
    ]
    .isna()
    .any(axis=1)
    .sum()
)

print(
    "Candidates with missing coordinates:",
    missing_coordinates
)

# ------------------------------------------------------------
# 9. Anomaly score integrity
# ------------------------------------------------------------

missing_scores = (
    final_ml_output[
        "anomaly_score"
    ]
    .isna()
    .sum()
)

print(
    "Candidates with missing anomaly score:",
    missing_scores
)

# ------------------------------------------------------------
# 10. Date integrity
# ------------------------------------------------------------

invalid_dates = candidate_dates.isna().sum()

print(
    "Candidates with invalid dates:",
    invalid_dates
)

# ------------------------------------------------------------
# 11. Nearest-state integrity
# ------------------------------------------------------------

missing_nearest_state = (
    final_ml_output[
        "nearest_indian_state"
    ]
    .isna()
    .sum()
)

print(
    "Missing nearest Indian state:",
    missing_nearest_state
)

# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

validation_passed = (
    len(earthquake_ml) == 6790
    and duplicate_count == 0
    and not bool(temporal_leakage)
    and len(final_ml_output) == 22
    and missing_coordinates == 0
    and missing_scores == 0
    and invalid_dates == 0
    and missing_nearest_state == 0
)

print()
print("=" * 70)

if validation_passed:
    print("FINAL ML VALIDATION: PASSED")
else:
    print("FINAL ML VALIDATION: REVIEW REQUIRED")

print("=" * 70)
```
